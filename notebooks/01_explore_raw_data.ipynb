## 6. Ringkasan Temuan Awal

Dataset transaksi perbankan ini terdiri dari **200.000 baris dan 24 kolom**, yang merepresentasikan aktivitas transaksi nasabah pada berbagai cabang bank di India.

**Struktur data:**
- Terdapat **identifier unik** untuk entitas utama:
  - `Customer_ID` (unik per baris)
  - `Transaction_ID` (unik per transaksi)
  - `Merchant_ID` (unik per merchant)
- Data mencakup kombinasi informasi:
  - **Nasabah**: `Customer_Name`, `Gender`, `Age`, `State`, `City`
  - **Akun & cabang**: `Account_Type`, `Bank_Branch`
  - **Transaksi**: `Transaction_Date`, `Transaction_Time`, `Transaction_Amount`, `Transaction_Type`
  - **Merchant**: `Merchant_Category`
  - **Konteks transaksi**: `Transaction_Device`, `Device_Type`, `Transaction_Location`
  - **Indikasi risiko**: `Is_Fraud`

**Tipe data dan format:**
- Kolom numerik utama:
  - `Transaction_Amount` (rata-rata ±49.5K, rentang 10–98K)
  - `Account_Balance` (rata-rata ±52.4K)
  - `Age` (rentang 18–70)
- Kolom target `Is_Fraud` bersifat biner (0/1) dengan proporsi fraud sekitar **5%**, menunjukkan **class imbalance** yang perlu diperhatikan pada analisis lanjutan.
- Kolom tanggal (`Transaction_Date`) dan waktu (`Transaction_Time`) masih berbentuk **string**, sehingga perlu parsing ke tipe datetime pada tahap technical cleaning.

**Kualitas data mentah:**
- Tidak ditemukan **missing values** pada seluruh kolom.
- Tidak terdapat duplikasi baris secara eksplisit.
- Beberapa kolom bersifat **high-cardinality**, seperti `Customer_Name`, `Customer_Email`, dan `Customer_Contact`, yang perlu dipertimbangkan penggunaannya pada tahap analisis dan visualisasi.
- Kolom `Transaction_Currency` hanya memiliki satu nilai unik (`INR`), sehingga secara analitis tidak memberikan variasi.

**Kesimpulan awal:**
Dataset ini **lengkap secara teknis** dan siap untuk masuk ke tahap **technical data cleaning menggunakan Python**, khususnya untuk:
- standarisasi nama kolom,
- konversi tipe data tanggal & waktu,
- penyiapan skema yang konsisten sebelum dimuat ke PostgreSQL sebagai raw table.
